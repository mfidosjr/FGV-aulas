## 1. Onde as Redes Neurais Fazem Diferença na Prática

**Por que usar uma rede neural em vez de um modelo de Machine Learning "tradicional"?**

Modelos tradicionais de ML (regressão linear/logística, árvores de decisão, random forest, SVM, k-means) continuam sendo ótimas escolhas para muitos problemas -- geralmente são mais rápidos de treinar, mais baratos de rodar em produção e mais fáceis de explicar. A rede neural passa a fazer diferença quando o problema tem características que esses modelos têm dificuldade de capturar:

- **Dados não estruturados**: imagem, áudio, texto livre e vídeo. Modelos tradicionais normalmente exigem que um especialista crie manualmente as variáveis de entrada (feature engineering); redes neurais aprendem essas representações sozinhas, direto dos dados brutos.
- **Relações muito não lineares e com muitas interações entre variáveis**: como vimos na Aula 2, cada camada oculta adicional permite representar combinações mais complexas do que uma equação linear ou uma árvore rasa conseguem.
- **Grandes volumes de dados**: modelos tradicionais tendem a "saturar" o desempenho depois de um certo volume de dados; redes neurais (principalmente as profundas) costumam continuar melhorando à medida que mais dados e mais poder computacional são adicionados.
- **Tarefas de ponta em visão computacional, linguagem natural e fala**, onde redes neurais (CNNs, RNNs/LSTMs, Transformers) hoje são o estado da arte.

Em troca, redes neurais exigem mais dados, mais poder computacional, mais tempo de ajuste, e são mais difíceis de explicar -- o trade-off de explicabilidade que já vimos na tabela comparativa da Aula 2 (redes neurais vs. equações/regras manuais). Ou seja, a escolha entre "ML tradicional" e "rede neural" é uma decisão de custo-benefício, não uma hierarquia onde uma é sempre melhor que a outra.

### Tipos de função de ativação

Já conhecemos três funções de ativação: o degrau (Aula 1, perceptron), e a ReLU e a sigmoid (Aula 2). A tabela abaixo situa essas três ao lado de outras comuns em redes mais profundas:

| **Função de Ativação** | **Intervalo de Saída** | **Uso Principal** | **Vantagens** | **Desvantagens** |
|----------------|----------------|----------------|----------------|----------------|
| **Degrau (Step Function)** | {0,1} | Perceptron simples | Fácil de entender | Apenas resolve problemas linearmente separáveis |
| **Sigmoid** | (0,1) | Classificação binária | Interpretação probabilística | Vanishing gradient |
| **Tanh** | (-1,1) | Classificação binária | Melhor que sigmoid, saída centrada | Ainda sofre de vanishing gradient |
| **ReLU** | [0,∞) | Redes profundas | Resolve vanishing gradient | Dying ReLU |
| **Leaky ReLU** | (-∞,∞) | Redes profundas | Evita dying ReLU | Requer ajuste do coeficiente |
| **Softmax** | (0,1) | Classificação multiclasse | Transforma em probabilidades | Computacionalmente caro |

## 3. Por que usar um framework?

Na Aula 2, cada peso da rede foi escolhido por nós manualmente. Na demonstração de quebra de simetria, chegamos a *treinar* pesos de verdade, mas usando a regra do perceptron -- que só sabe ajustar os pesos de **uma única camada**, comparando a saída diretamente com o rótulo.

Numa rede com camada oculta, o problema fica mais difícil: como saber o quanto cada peso da camada oculta contribuiu para o erro final, se o erro só é calculado na camada de saída? A resposta é o **backpropagation** (RUMELHART; HINTON; WILLIAMS, 1986): um algoritmo que propaga o erro da camada de saída de volta para as camadas anteriores, calculando o quanto cada peso -- de cada camada -- deveria mudar. É, na prática, a mesma ideia da regra do perceptron (ajustar o peso proporcionalmente ao erro), só que generalizada matematicamente para redes com várias camadas.

Implementar backpropagation à mão para uma rede grande seria trabalhoso e propenso a erros. É exatamente isso que frameworks como **Keras** (sobre o **TensorFlow**) automatizam: você descreve a arquitetura (camadas, neurônios, ativações) e o framework calcula e aplica as atualizações de peso de cada camada automaticamente, a cada época de treino.

## 4. Revisitando o e-mail de spam com Keras

Vamos usar exatamente o mesmo problema e a mesma arquitetura da Aula 2 (3 entradas → 3 neurônios ocultos com ReLU → 1 neurônio de saída com sigmoid) -- mas, desta vez, em vez de escolher os pesos à mão, vamos deixar o Keras aprendê-los a partir de dados, usando backpropagation.

Para isso, também precisamos de mais dados do que os 10 e-mails da Aula 1: com poucos exemplos, qualquer rede "decora" o treino em vez de aprender um padrão generalizável. Vamos gerar uma base sintética maior (600 e-mails), com as mesmas 3 características (links, linguagem suspeita, confiabilidade do remetente), e separar uma parte para teste -- e-mails que a rede nunca viu durante o treino -- para medir a acurácia de forma honesta.

```python
import numpy as np
from sklearn.model_selection import train_test_split

seed = 42
np.random.seed(seed)

n_por_classe = 300

# E-mails de spam: muitos links, linguagem suspeita, remetente pouco confiavel
spam = np.column_stack([
    np.clip(np.random.normal(7, 2, n_por_classe), 0, 10),
    np.clip(np.random.normal(7, 2, n_por_classe), 0, 10),
    np.clip(np.random.normal(2, 1.5, n_por_classe), 0, 10),
])

# E-mails legitimos: poucos links, linguagem normal, remetente confiavel
nao_spam = np.column_stack([
    np.clip(np.random.normal(2, 1.5, n_por_classe), 0, 10),
    np.clip(np.random.normal(2, 1.5, n_por_classe), 0, 10),
    np.clip(np.random.normal(7, 2, n_por_classe), 0, 10),
])

X = np.vstack([spam, nao_spam])
y = np.concatenate([np.ones(n_por_classe), np.zeros(n_por_classe)])

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.2, random_state=seed, stratify=y
)

print('Total de e-mails simulados:', len(X))
print('Exemplos de treino:', len(X_treino))
print('Exemplos de teste:', len(X_teste))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

tf.random.set_seed(seed)

modelo = Sequential()
modelo.add(Input(shape=(3,)))
modelo.add(Dense(3, activation='relu', name='CamadaOculta'))
modelo.add(Dense(1, activation='sigmoid', name='CamadaSaida'))

modelo.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("Treinando a rede neural...")
historico = modelo.fit(X_treino, y_treino, epochs=100, verbose=0)

perda_teste, acuracia_teste = modelo.evaluate(X_teste, y_teste, verbose=0)
print(f"Acuracia no conjunto de teste (e-mails nunca vistos no treino): {acuracia_teste:.2%}")
```

### Explicação do Código

Cada peça do Keras corresponde a algo que já vimos nas Aulas 1 e 2:

- **`Sequential`**: uma pilha de camadas, o mesmo fluxo entrada → oculta → saída que já usamos.
- **`Dense(3, activation='relu')`**: a camada oculta com 3 neurônios e ReLU, igual à da Aula 2 -- só que agora os pesos começam **aleatórios** (a quebra de simetria da Aula 2, aplicada automaticamente) e são ajustados durante o treino, em vez de fixados à mão.
- **`Dense(1, activation='sigmoid')`**: a camada de saída, igual à da Aula 2.
- **`compile(optimizer='adam', loss='binary_crossentropy')`**: escolhe o algoritmo que ajusta os pesos a cada passo (uma versão mais sofisticada de `peso = peso + taxa_aprendizado * erro * entrada`, a mesma regra da Aula 1) e a forma de medir o erro para um problema de classificação binária.
- **`fit(X_treino, y_treino, epochs=100)`**: roda o backpropagation por 100 épocas -- o mesmo conceito de "época" da Aula 1.
- **`evaluate` / `predict`**: a mesma inferência que já calculamos manualmente na Aula 2, agora automática.

```python
email_exemplo = np.array([[8.0, 7.0, 2.0]])  # o mesmo e-mail de exemplo desde a Aula 1
probabilidade = modelo.predict(email_exemplo, verbose=0)
print(f"Probabilidade de spam para o e-mail de exemplo [links=8, linguagem_suspeita=7, confiabilidade=2]: {probabilidade[0][0]:.2%}")
```

Na Aula 2, com pesos escolhidos manualmente, esse mesmo e-mail recebeu uma probabilidade de spam de aproximadamente 83%. Agora, com pesos **aprendidos** a partir de 480 exemplos de treino via backpropagation, a rede chega a uma conclusão semelhante -- ambos os caminhos (pesos manuais bem escolhidos e pesos aprendidos de dados reais) convergem para a mesma direção qualitativa, o que é uma boa checagem de sanidade: a intuição que construímos à mão nas Aulas 1 e 2 é consistente com o que a rede aprende sozinha.

## 5. O mesmo exemplo com PyTorch

Keras não é o único framework popular para treinar redes neurais -- o **PyTorch** é outra escolha muito usada, especialmente em pesquisa. A diferença mais visível entre os dois não é matemática, é de **estilo**:

- No Keras, `compile()` e `fit()` escondem o laço de treinamento: você descreve o que quer, e o framework decide como rodar cada época.
- No PyTorch, você escreve o laço de treinamento **explicitamente**: a cada lote de dados, zera os gradientes, calcula a saída, calcula o erro, propaga o erro para trás (`backward()`) e atualiza os pesos (`step()`) -- os mesmos passos que o Keras faz por trás das cortinas.

Vamos treinar exatamente a mesma arquitetura (3 entradas → 3 neurônios ocultos com ReLU → 1 neurônio de saída com sigmoid), usando o **mesmo conjunto de treino e teste** gerado na seção anterior, para comparar os dois frameworks de forma justa.

```python
import torch
import torch.nn as nn

torch.manual_seed(seed)

X_treino_t = torch.from_numpy(X_treino.astype('float32'))
y_treino_t = torch.from_numpy(y_treino.astype('float32')).unsqueeze(1)
X_teste_t = torch.from_numpy(X_teste.astype('float32'))
y_teste_t = torch.from_numpy(y_teste.astype('float32')).unsqueeze(1)

modelo_pt = nn.Sequential(
    nn.Linear(3, 3),
    nn.ReLU(),
    nn.Linear(3, 1),
    nn.Sigmoid(),
)

criterio = nn.BCELoss()
otimizador = torch.optim.Adam(modelo_pt.parameters())

batch_size = 32
n_treino = X_treino_t.shape[0]

print("Treinando a rede neural (PyTorch)...")
for epoca in range(100):
    permutacao = torch.randperm(n_treino)
    for inicio in range(0, n_treino, batch_size):
        idx_lote = permutacao[inicio:inicio + batch_size]
        lote_X, lote_y = X_treino_t[idx_lote], y_treino_t[idx_lote]

        otimizador.zero_grad()
        saida = modelo_pt(lote_X)
        perda = criterio(saida, lote_y)
        perda.backward()
        otimizador.step()

with torch.no_grad():
    pred_teste = modelo_pt(X_teste_t)
    acuracia_pt = ((pred_teste > 0.5).float() == y_teste_t).float().mean().item()
    print(f"Acuracia no conjunto de teste (PyTorch): {acuracia_pt:.2%}")

    email_exemplo_pt = torch.tensor([[8.0, 7.0, 2.0]])
    probabilidade_pt = modelo_pt(email_exemplo_pt).item()
    print(f"Probabilidade de spam para o e-mail de exemplo (PyTorch): {probabilidade_pt:.2%}")
```

### Explicação do Código

- `nn.Sequential(nn.Linear(3,3), nn.ReLU(), nn.Linear(3,1), nn.Sigmoid())`: a mesma arquitetura da versão Keras, só com outra sintaxe -- `nn.Linear` é o equivalente do `Dense`.
- `nn.BCELoss()` e `torch.optim.Adam(...)`: o mesmo par perda/otimizador do Keras, mas instanciados separadamente em vez de passados para um `compile()`.
- O laço `for epoca in range(100): for inicio in range(0, n_treino, batch_size): ...` faz manualmente o que `model.fit(..., epochs=100)` fazia sozinho no Keras -- inclusive dividir os dados em lotes (`batch_size=32`, o mesmo padrão do Keras).
- `perda.backward()` é o passo de backpropagation explícito: o PyTorch calcula o gradiente do erro em relação a cada peso da rede.
- `otimizador.step()` aplica a atualização de pesos usando esses gradientes -- o mesmo `peso = peso + ajuste` que já vimos, calculado automaticamente pelo Adam.

**Comparando os dois frameworks:** com a mesma arquitetura e os mesmos dados de treino/teste, Keras e PyTorch chegaram a resultados de acurácia muito próximos -- o que é esperado, já que ambos implementam exatamente o mesmo algoritmo (backpropagation com Adam). A diferença prática entre eles é de **controle e verbosidade**: o Keras é mais rápido de escrever para casos comuns; o PyTorch dá mais controle explícito sobre cada etapa do treino, o que é útil para arquiteturas customizadas ou pesquisa.
